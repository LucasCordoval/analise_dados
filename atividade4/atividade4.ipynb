```python
import pandas as pd
```

```python
caminho = r"H:\python\analise_de_dados\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
tabela = pd.read_excel(caminho, header=2)
tabela
```

```python
dados_trabalhados = tabela.iloc[0:5570,1:]
```

```python
dados_trabalhados["diferença"] = (dados_trabalhados["População Censo 2022"] - dados_trabalhados["População Município 2010\n(Sinopse)"])
dados_trabalhados
```

```python
dados_trabalhados.columns
```

```python
porestado = (
    dados_trabalhados.groupby('UF')[[
        'População Município 2010\n(Sinopse)',
       'População 2010 (Alterações de Limites até 2022)1',
       'População Censo 2022', 'diferença']]
    .sum()
    .reset_index()
)
porestado   
```

```python
porestado = porestado.sort_values(by='diferença', ascending=False)
porestado
```

```python
porestado.to_csv(r"H:\python\analise_de_dados\Ctabelaminhaatividade4.csv", sep =";")
```

```python
pormunicipio = (
    dados_trabalhados.groupby('NOME DO MUNICÍPIO')[[
        'População Município 2010\n(Sinopse)',
       'População 2010 (Alterações de Limites até 2022)1',
       'População Censo 2022', 'diferença']]
    .sum()
    .reset_index()
)
pormunicipio
```

```python
pormunicipi = tabela.iloc[0:5570,3:]
pormunicipi["diferença"] = (pormunicipi["População Censo 2022"] - pormunicipi["População Município 2010\n(Sinopse)"])
pormunicipi
```

```python
pormunicipi = pormunicipi.sort_values(by='diferença', ascending=False)
pormunicipi
```

```python
pormunicipi.to_csv(r"H:\python\analise_de_dados\Ctabelaminhaatividade4municipio.csv", sep =";")
```
